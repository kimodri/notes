# Introduction to Pytorch

## Tensors

Pytorch run calculations on tensors. That is it. You can get tensors either with numpy or python list:

```python
import torch
import numpy as np

sample_tensor = [[1, 2, 3,]]  # Tensor with 3 elements
print(type(sample_tensor))

torch_tensor = torch.tensor(sample_tensor)
print(f"Shape: {torch_tensor.shape}")
print(f"Data types: {torch_tensor.dtype}")
print(type(torch_tensor))
```

## Neural Network

A neural network is built with `input`, `hidden`, and `output` layers.

### Simple Linear Network

Let's start with no hidden layer: `3 inputs -> 2 outputs`.

![Three-input, two-output linear network](diagrams/00-simple-linear-network.svg)

Use `nn.Linear(3, 2)` for this.

```python
import torch
from torch import nn

torch.manual_seed(42)

linear_network = nn.Linear(in_features=3, out_features=2)
sample_input = torch.tensor([[1.0, 2.0, 3.0]])
output = linear_network(sample_input)

print(f"Input shape: {sample_input.shape}")
print(f"Output shape: {output.shape}")
print(f"Output: {output}")
```

## Hidden Layers

Hidden layers help the network learn complex patterns. `ReLU` adds non-linearity.

![Neural network with a hidden layer and ReLU](diagrams/00-hidden-layers.svg)

`nn.Sequential` stacks the layers in order: `3 -> 4 -> 2`.

**Parameter count for a linear layer:**

$$\text{parameters} = (\text{input features} \times \text{output features}) + \text{output features}$$

For this model:

- First layer: $(3 \times 4) + 4 = 16$ parameters
- ReLU: $0$ parameters
- Second layer: $(4 \times 2) + 2 = 10$ parameters
- Total: $16 + 10 = 26$ parameters

```python
hidden_network = nn.Sequential(
    nn.Linear(in_features=3, out_features=4),
    nn.ReLU(),
    nn.Linear(in_features=4, out_features=2),
)

output = hidden_network(sample_input)
print(hidden_network)
print(f"Output shape: {output.shape}")
print(f"Output: {output}")

total_parameters = sum(parameter.numel() for parameter in hidden_network.parameters())
print(f"Total parameters: {total_parameters}")
```
