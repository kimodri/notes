## 1. INNER JOIN
- Only returns when there is a match.

```python
-- Example: Basic INNER JOIN to match records by the "id" field
SELECT left_table.id, left_table.left_val, right_table.right_val
FROM left_table
INNER JOIN right_table
ON left_table.id = right_table.id;
```

```python
import pandas as pd   
from IPython.display import display_html 

df1 = pd.DataFrame({"id": [1, 2, 3, 4], "left_val": ['L1','L2', 'L3', 'L4']})
df2 = pd.DataFrame({"id": [1, 4, 5, 6], "right_val": ['R1', 'R2', 'R3', 'R4']})
df3 = pd.DataFrame({"id": [1, 4], "left_val": ['L1', 'L4'], "right_val": ['R1', 'R2']})

df1_styler = df1.style.set_table_attributes("style='display:inline'").set_caption('Left table')
df2_styler = df2.style.set_table_attributes("style='display:inline'").set_caption('right table')
df3_styler = df3.style.set_table_attributes("style='display:inline'").set_caption('Result')
display_html(df1_styler._repr_html_()+df2_styler._repr_html_()+df3_styler._repr_html_(), raw=True)
```

### 1.1. USING() Clause
- When the field names are the same in both tables, you can use the `USING` clause instead of `ON` to simplify the query.

```python
-- Example: Using the USING clause for a simpler INNER JOIN
SELECT p1.country, p1.continent, p2.prime_minister
FROM world.presidents AS p1
INNER JOIN world.prime_ministers AS p2
USING (country);
```

### 1.2. Multiple JOINS
- SQL allows you to combine multiple joins in a single query to retrieve data from several tables at once.
- You can join one table, then chain another based on the result of the first join.

```python
-- let this be the first JOIN
-- joining countries with both prime ministers and presidents
SELECT 
	p1.country AS countries, 
	p1.continent AS continents, 
	p2.president AS presidents,
	p1.prime_minister AS prime_ministers
FROM world.prime_ministers AS p1
INNER JOIN world.presidents AS p2
ON p1.country = p2.country;
```

- in that result will the next table that we'll join will depend on.

```python
SELECT 
	p1.country AS countries, 
	p1.continent AS continents, 
	p2.president AS presidents,
	p1.prime_minister AS prime_ministers,
	p3.pm_start
FROM world.prime_ministers AS p1
INNER JOIN world.presidents AS p2
ON p1.country = p2.country
-- the table below will depend on the table above
INNER JOIN world.prime_minister_terms AS p3
ON p1.prime_minister = p3.prime_minister;
```

**Explanation:**
- First, an INNER JOIN is performed between the presidents and prime_ministers tables based on the country field.
- Then, another INNER JOIN is performed between the result of the first join and the prime_minister_terms table, based on the prime_minister field.

## 2. LEFT and RIGHT JOIN 
- Allowing us to include non-matching records from one of the tables involved in the join.

### 2.1. The following visualizes the left `JOIN`

```python
SELECT id, left_table.left_val, right_table.right_val
FROM left_table
LEFT JOIN right_table
ON left_table.id = right_table.id;
```

```python
import pandas as pd   
from IPython.display import display_html 

df1 = pd.DataFrame({"id": [1, 2, 3, 4], "left_val": ['L1','L2', 'L3', 'L4']})
df2 = pd.DataFrame({"id": [1, 4, 5, 6], "right_val": ['R1', 'R2', 'R3', 'R4']})
df3 = pd.DataFrame({"id": [1, 2, 3, 4], "left_val": ['L1', 'L2', 'L3', 'L4'], "right_val": ['R1', 'NaN', 'NaN','R2']})

df1_styler = df1.style.set_table_attributes("style='display:inline'").set_caption('Left table')
df2_styler = df2.style.set_table_attributes("style='display:inline'").set_caption('right table')
df3_styler = df3.style.set_table_attributes("style='display:inline'").set_caption('Result')
display_html(df1_styler._repr_html_()+df2_styler._repr_html_()+df3_styler._repr_html_(), raw=True)
```

## 4. CROSS JOIN
- A `CROSS JOIN` produces all possible combinations of the records from both tables.
- It has **NO** `ON`

```python
-- The following visualizes the use of CROSS JOIN
SELECT table1.id, table2.id
FROM table1
CROSS JOIN table2;
```

## 5. SELF JOINS
- A self `JOIN` is when a table is joined with itself. It is used to compare rows within the same table.
- Aliasing is required

If for example, we want to _pair_ countries of the same continent, we can do this:

```python
SELECT p1.country AS country1, p2.country AS country2, p1.continent
FROM world.prime_ministers AS p1
INNER JOIN world.prime_ministers AS p2
ON p1.continent = p2.continent
-- WHERE p1.country != p2.country
-- or
-- AND p1.country != p2.country
LIMIT 10;
```

Notice that coutries paired up with themselves, you can change that with a `WHERE` or the `AND` clause.

## 2. INTERSECT
- `INTERSECT` is a set operation that takes two tables as input and returns only the records that exist in both tables.

### 2.1. Intersect vs Inner Join
- **Since `INTERSECT` is a set operator**: `INTERSECT` requires all fields in both tables to match, and the tables must have the same number of columns.
- `INNER JOIN` adds more columns and may return duplicate records, while INTERSECT only returns unique common records.

```python
import pandas as pd   
from IPython.display import display_html 

df1 = pd.DataFrame({"id": [1, 1, 3, 4], "val": ['N1','N1', 'L3', 'L4']})
df2 = pd.DataFrame({"id": [1, 4, 5, 6], "val": ['N1', 'R2','R3', 'R4']})
df3 = pd.DataFrame({"id": [1], "left_val": ['N1']})

df1_styler = df1.style.set_table_attributes("style='display:inline'").set_caption('Left table')
df2_styler = df2.style.set_table_attributes("style='display:inline'").set_caption('right table')
df3_styler = df3.style.set_table_attributes("style='display:inline'").set_caption('Result')
display_html(df1_styler._repr_html_()+df2_styler._repr_html_()+df3_styler._repr_html_(), raw=True)
```

## 1. Subquerying w/ Semi Joins and Anti Joins

ANKI THIS ONE:
- Unlike "additive" joins, which add columns to the result set, semi and anti joins **filter** records based on conditions in a related table.

### 1.1. Semi Joins
- A semi join retrieves records from the left table where a specified condition is met in the right table.
- Dapat lalabas sa result yung mga nandoon sa right table. We will use `IN` in here.

**for example**: Let's say I want to filter for countries that gain independence before 1800, I can do that with just one table containg the country and the independence year column using WHERE. But what if I also want to have the presidents thus I need another table so it is like a JOIN but instead of adding we are filtering.

```python
-- example:
-- perhaps anki this
SELECT *
FROM world.countries
LIMIT 10;
```

```python
-- we can simply do it with this if we will be just using one table
SELECT world.countries.name
FROM world.countries
WHERE world.countries.indep_year < 1800
LIMIT 10;
```

Although the above code can be sufficient it still lacks tha column for the presidents and too bad that table does not have a president column therefore it only makes sense to use another.

```python
-- This is where we can get the presidents column
SELECT world.presidents.country, world.presidents.president
FROM world.presidents
LIMIT 10;
```

```python
-- But then remember that we are also tasked to add the world leaders so we can do this
SELECT world.presidents.country, world.presidents.president
FROM world.presidents

WHERE world.presidents.country IN (
SELECT world.countries.name
FROM world.countries 
WHERE world.countries.indep_year < 1800) -- remember that this query here returns what's above

LIMIT 10;
```

### 1.2. Anti Join
- An anti join returns records in the left table that do not match any records in a specified column of the right table.

```python
-- okay, now the opposite we want to return values that are NOT IN the second table
-- the task is to return countries not in the second table (query)
-- we want to return countries name whose basic_unit is Peso
```

```python
-- let's generate the first table first
SELECT *
FROM world.currencies;
```

```python
-- we'll filter out countries whose unit is not peso 
SELECT world.currencies.code
FROM world.currencies
WHERE world.currencies.basic_unit NOT LIKE '%peso'
LIMIT 10;
```

```python
-- now, we can use that table to filter countries that use Peso, in a way it's like asking what are the countries not in the above table
SELECT world.countries.code, world.countries.name
FROM world.countries

WHERE world.countries.code NOT IN (SELECT world.currencies.code
FROM world.currencies
WHERE world.currencies.basic_unit NOT LIKE '%peso')
LIMIT 10;
```

## 2. Subqueries within WHERE and SELECT

### 2.1. Subqueries inside where
- This is the same as what we did with Semi and Anti Joins and thus used in filtering

To identify what we need to do let us generate the tables.

```python
SELECT *
FROM world.populations
LIMIT 5;
```

```python
-- what if we want countries and continent with life expectanct of lower than 50
SELECT world.countries.name, world.countries.continent
FROM world.countries
WHERE world.countries.code IN (SELECT world.populations.country_code
							  FROM world.populations
							  WHERE world.populations.life_expectancy < 50)
LIMIT 10;
```

### 2.2. Subqueries with SELECT


- If the subquery inside `WHERE` is used for filtering subquery inside `SELECT` clause can calculate aggregated values dynamically as part of each row in the main query.
- It requires an allias which makes sense because with `SELECT` you are returning a column
