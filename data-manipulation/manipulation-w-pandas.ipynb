## 2. Custome Aggregation with .agg()
- Use `.agg()` to apply custom functions across DataFrame columns.
- Create a custom function for specific calculations, such as the 30th percentile.
- `df['col'].agg(func)`

````python
# Define a custom function to calculate 30th percentile
def pct30(column):
    return column.quantile(0.3)

# Applying custom function to weight column
thirtieth_percentile_weight = dogs['weight_kg'].agg(pct30)
thirtieth_percentile_weight
````

**Aggregating Multiple Columns and Functions**
- With `.agg(),` apply multiple functions on multiple columns at once.
- **Example:** Calculate both the 30th and 40th percentiles.

````python
# Define additional custom functions
def pct40(column):
    return column.quantile(0.4)

# Apply multiple custom functions on weight and height columns
dogs[['weight_kg', 'height_cm']].agg([pct30, pct40])
````

````python
# you can also add functions from numpy
dogs[['weight_kg', 'height_cm']].agg([pct30, pct40, np.median])
````

### 5.6. Multi-Column Grouping w/ Multi-Column Aggregation
- .groupby() combined with .agg() can handle multiple grouping variables and multiple summary statistics.

````python
import pandas as pd

# Sample data
data = {
    'Category': ['A', 'A', 'A', 'B', 'B', 'C', 'C', 'C'],
    'Subcategory': ['X', 'Y', 'X', 'Y', 'X', 'Y', 'X', 'X'],
    'Value': [10, 20, 30, 40, 50, 60, 70, 80],
    'Quantity': [1, 2, 1, 3, 2, 1, 2, 3]
}
df = pd.DataFrame(data)
print(df)
# Group by 'Category' and 'Subcategory' and apply multiple aggregatis
grouped = df.groupby(['Category', 'Subcategory']).agg(['sum', 'mean', 'count']).reset_index()
print(grouped)
````

## 6. Pivot Tables
- The .pivot_table() method can replace .groupby() for simple grouped statistics.
- When does your pivot value return a NaN if for example in avocado: your conventional does not have a small size but the column has a small size

### 6.1. Basic Pivot Tables
- Use the `values` argument for the column you want to summarize and `index` for the column to group by. By default, `.pivot_table()` computes the mean for each group.

````python
# remember that you can do this with groupby()
print(type(avocado.groupby('type')['nb_sold'].mean()))
print(avocado.groupby('type')['nb_sold'].mean())
````

````python
# you can also do it with pivot tables
pvTableAvocado = avocado.pivot_table(index = 'type', values = 'nb_sold')
print(type(pvTableAvocado))
print(pvTableAvocado)
````

### 6.3. Customizing Aggregation Function
- To use a different statistic, pass the desired function to the `aggfunc` argument.

````python
# customizing aggregation function
# if you want the above code to be a median instead of a mean
print(avocado.pivot_table(index = 'type', values = 'nb_sold', aggfunc = np.median))
````

### 6.4. Multiple Statistics
- For multiple statistics, pass a list of functions to `aggfunc`.

````python
# If you want to calculate mean and median of the type of avocados
print(avocado.pivot_table(index = 'type', values = 'nb_sold', aggfunc = [np.median, np.mean]))
````

### 3.1. Slicing the Outer Index Level with .loc

````python
print("This one will work!")
print(temperatures_srt.loc[('Pakistan', 'Lahore'):('Russia', 'Moscow')])
````
